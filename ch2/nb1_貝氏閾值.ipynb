# 📓 Notebook 1｜一維貝氏閾值：重現課本 Example 2.1

> 對應講義 Part 1 ＋ Demo 1。驗證：① 最小錯誤閾值 x0=1/2；② 加入損失矩陣後閾值左移。

```
import numpy as np
import matplotlib.pyplot as plt
S2, M1, M2 = 0.5, 0.0, 1.0          # 兩類高斯方差與均值（課本 2.1）
g = lambda x, m: np.exp(-(x-m)**2/(2*S2))/np.sqrt(2*np.pi*S2)

def Pe_at(x0, N=2001):
    x = np.linspace(-2, 3, N); h = x[1]-x[0]
    L = x < x0
    return (0.5*g(x, M2)*L + 0.5*g(x, M1)*~L).sum()*h   # 兩塊誤差面積

x0s = np.linspace(-1, 2, 501)
pes = np.array([Pe_at(a) for a in x0s])
best = x0s[np.argmin(pes)]
print(f'最小錯誤閾值 x0 = {best:.4f}  （理論 0.5）')
print('Pe(x0=0.5) =', round(Pe_at(0.5), 5))

plt.figure(figsize=(9, 4))
plt.plot(x0s, pes)
plt.axvline(0.5, color='r', ls='--', label='x0=1/2（最小錯誤）')
plt.axvline((1-np.log(2))/2, color='g', ls='--', label='(1−ln2)/2（最小風險）')
plt.xlabel('閾值 x0'); plt.ylabel('錯誤機率 Pe'); plt.legend(); plt.title('課本 Example 2.1：閾值最優性')
plt.show()
```

```
# 損失矩陣 λ12=0.5, λ21=1.0 → 最小化 r(θ)
def risk_at(x0, lam12=0.5, lam21=1.0, N=2001):
    x = np.linspace(-2, 3, N); h = x[1]-x[0]
    L = x < x0
    return (lam21*0.5*g(x, M1)*L + lam12*0.5*g(x, M2)*~L).sum()*h + 0   # +0.5(λ11+λ22零)

x0r = np.linspace(-1, 2, 1001)
rs = np.array([risk_at(a) for a in x0r])
best_r = x0r[np.argmin(rs)]
print(f'最小風險閾值 = {best_r:.4f}  （理論 (1−ln2)/2 ≈ {(1-np.log(2))/2:.4f}）')
print('→ 誤判 ω2 較嚴重 → 閾值左移、ω2 決策區擴大 ✅')
```
