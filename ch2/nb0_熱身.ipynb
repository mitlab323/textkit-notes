# 📓 Notebook 0｜熱身：高斯與貝氏定理（15 分鐘）

> 對應講義 Part 1。目標：會用 NumPy 抽高斯、會算貝氏後驗。

```
import numpy as np
def gauss(x, mu=0.0, sigma2=1.0):
    return np.exp(-(x-mu)**2/(2*sigma2))/np.sqrt(2*np.pi*sigma2)

xs = np.linspace(-3, 3, 1001)
print('高斯 N(0,1) 在 x=0 的值:', round(gauss(0.0), 5))

# 貝氏定理：似然 × 先驗 ÷ p(x)
lik = [gauss(1.0, 0.0, 1.0), gauss(1.0, 2.0, 1.0)]      # 兩類在 x=1 的似然
prior = [0.7, 0.3]
post = np.array(lik) * prior; post = post / post.sum()  # 後驗
print('後驗 P(ωᵢ|x=1):', np.round(post, 3), '→ 應選 ω1')
```

> ✏️ 改 `prior` 為 [0.2, 0.8]：後驗會翻盤嗎？——這就是「先驗的威力」。
