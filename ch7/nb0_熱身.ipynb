### NumPy 五分鐘速成

`np.array` 是我們的主角——它就是一張「圖片」的靈魂：二維陣列 = 灰階影像。

```
# 一張 4×4 的「假圖片」（0=黑, 3=白）
img = np.array([
    [0, 0, 2, 2],
    [1, 1, 0, 0],
    [3, 2, 3, 3],
    [3, 2, 2, 2],
])
print('形狀 (高, 寬):', img.shape)
print('最大值:', img.max(), ' 最小值:', img.min())
print('第 2 列:', img[1])          # 索引從 0 開始
print('第 1 列第 2 格:', img[0, 2])
```

```
# 顯示圖片：灰階 = imshow(..., cmap='gray', vmin=0, vmax=3)
plt.figure(figsize=(2.5, 2.5))
plt.imshow(img, cmap='gray', vmin=0, vmax=3)
plt.colorbar(shrink=0.8)
plt.title('一張 4×4 影像')
plt.axis('off')
plt.show()
```
