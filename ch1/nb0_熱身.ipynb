### 重點：二維陣列 = 「一筆資料」

模式辨認裡，我們把每個樣本表示成一列數字（特徵向量）；一堆樣本疊起來就是一個**矩陣**。

```
# 5 個樣本 × 2 個特徵（例如：每個樣本的 mean 與 std）
X = np.array([
    [3.2, 1.1],
    [3.5, 1.4],
    [7.1, 1.9],
    [6.8, 1.6],
    [4.0, 3.5],
])
print('形狀 (樣本數, 特徵數):', X.shape)
print('第 3 個樣本的特徵向量 x =', X[2])
```

```
x = X[:, 0]; y = X[:, 1]
plt.figure(figsize=(5, 4))
plt.scatter(x, y, s=60)
plt.xlabel('特徵 x1（mean）'); plt.ylabel('特徵 x2（std）')
plt.title('特徵空間裡的 5 個樣本')
plt.show()
```
